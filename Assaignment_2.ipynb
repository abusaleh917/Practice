```python
#H1 Write a function to calculate the sum of numbers from 1 to N.
print("Here is a function fot calculate the sum of numbers 1 to N:")
def summ(n):
    res=0

    for i in range(1, n+1):
      res = res+i

    return res

n=int(input("Enter the number:"))

print(summ(n))

```

```python
#H2 Write a function that returns the square of a number.
def sqrt(n):

    n=n*n
    return n

n=int(input("Enter the number:"))
print("The squrare of", n, "is", sqrt(n))
```

```python
#H3 Write a function that checks whether a number is even/odd.
def evod(n):

    if n%2==0:
      return "EVEN"

    if n%2!=0:
      return "ODD"

n=int(input("Enter the number:"))
print("The number", n, "is",evod(n))
```

```python
#H4 Write a function that returns the larger number from two given numbers.
def large_num(num1,num2):

    if num1>num2:
      print("The number",num1, "is larger than",num2)

    elif num2>num1:
      print("The number",num2, "is larger than",num1)


    else:
      print("The numbers are equal")

num1=int(input("Enter the first number:"))
num2=int(input("Enter the second number:"))
large_num(num1,num2)
```
