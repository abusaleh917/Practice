```python
#H1 Write a function to calculate the sum of numbers from 1 to N.
print("Here is a function fot calculate the sum of numbers 1 to N:")
def summ(n):
    res=0

    for i in range(1, n+1):
      res = res+i

    return res

n=int(input("Enter the number:"))

result=summ(n)

print(result)

```
