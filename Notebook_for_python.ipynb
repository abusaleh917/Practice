```python
## H8: Create a program that converts temperatures between Fahrenheit and Celsius based on user input.
print("If your input is Fahrenheit please type ""F"" otherwise type C")
Con=(input())
if Con == "F" or Con=="f":
  print("Temperature in Fahrenheit=")
  temp=int(input())
  celsius=(temp - 32) * 5/9
  print("Temparature in celsius=", celsius)
elif Con == "C" or Con== "c":
  print("Temparature in celsius=")
  temp=int(input())
  fahrenheit = (temp * 9/5) + 32
  print("Temparature in Fahrenheit=", fahrenheit)
else:
    print("input is Invalid")
```

```python
# H11: Write a program that prompts the user to input a number N and calculates the sum of all numbers from 1 to N that are divisible by 3 or 5.
print("Enter a number N:")
N = int(input())
s=0


for i in range(0, N+1 , 1):
  if i%3==0 or i%5==0:
    s=s+i
print(s)
```
