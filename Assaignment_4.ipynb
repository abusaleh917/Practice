```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Sensible global defaults so every plot looks consistent
sns.set_theme(style='whitegrid', context='notebook')
plt.rcParams['figure.figsize'] = (7, 4)
plt.rcParams['figure.dpi']     = 600
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
gym      = pd.read_csv('/content/drive/MyDrive/Colab_Notebooks_Data_Science/gym_membership.csv')

print(gym.shape)
```

```python
#By using Matplotlib make a Bar chart of membership types — gym['abonoment_type'].value_counts()
membership_types = gym['abonoment_type'].value_counts()

fig, ax = plt.subplots()
ax.bar(membership_types.index, membership_types.values, color=['green', 'blue'])
ax.set(title='Membership Types', ylabel='Count')
plt.show()


```

```python
#Histogram of avg_time_in_gym
fig, ax = plt.subplots()
ax.hist(gym['avg_time_in_gym'], bins=10, color='skyblue', edgecolor='white')
ax.set(title='Average time in Gym',
       xlabel='avg_time_in_gym', ylabel='Frequency')
plt.show()
```

```python
#Scatter plot of Age vs visit_per_week
fig, ax = plt.subplots()
ax.scatter(gym['Age'], gym['visit_per_week'], alpha=0.4, s=18)
ax.set(title='Age vs visit_per_week',
       xlabel='Age', ylabel='visit_per_week')
plt.show()
```

```python
#By using Seaborn make Box plot — gender vs Age

fig, ax = plt.subplots()

sns.boxplot(data=gym, x='gender', y='Age', ax=ax)

ax.set_title('Age by Gender')

plt.tight_layout()

plt.show()
```

```python
#Count plot — days_per_week
fig, axes = plt.subplots()

sns.countplot(data=gym, x='days_per_week', ax=axes)
axes.set_title('days_per_week')

plt.tight_layout()
plt.show()
```

```python
#Bar plot — mean visit_per_week per abonoment_type
fig, ax = plt.subplots(figsize=(9, 4))
order = gym.groupby('abonoment_type')['visit_per_week'].mean().sort_values().index
sns.barplot(data=gym, x='abonoment_type',
            y='visit_per_week', order=order, ax=ax)
ax.set_title('Average visit per week by abonoment Type')
ax.tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.show()
```
