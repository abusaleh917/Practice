```python
#Code for One-Hot Encoding for the coloumn furnishingstatus(Codes from Gemini)
furnishingstatus_encoded_cols = pd.get_dummies(df['furnishingstatus'], prefix='furnishingstatus', drop_first=True, dtype=int)

df_with_one_hot = pd.concat([df.copy(), furnishingstatus_encoded_cols], axis=1)

print("Original 'furnishingstatus' values (first 5):")
print(df['furnishingstatus'].head())
print("\nDataFrame after One-Hot Encoding 'furnishingstatus' (first 5 rows with relevant columns):")
# Displaying only the original and new one-hot encoded columns for clarity
relevant_columns = ['furnishingstatus'] + [col for col in df_with_one_hot.columns if 'furnishingstatus_' in col]
print(df_with_one_hot[relevant_columns].head())
```

```python
# Encode the target variable 'y' ('airconditioning') as it's categorical (yes/no)
le_y = LabelEncoder()
y_encoded = le_y.fit_transform(y)
print(f"\nLabel Encoding for 'airconditioning' (y):")
print(dict(zip(le_y.classes_, le_y.transform(le_y.classes_))))

# Train-test split (80% train, 20% test) — remember stratify=y
# Use X_encoded for features and y_encoded for the target.
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y_encoded,
    test_size=0.2,
    stratify=y_encoded,
    random_state=30
)

print(f"\nWe will train with: {len(X_train)} houses")
print(f"We will test with: {len(X_test)} houses")
```

```python
# StandardScaler
scaler = StandardScaler()

# Get the original 'area' value of the first row before scaling X_train
# Assuming 'area' is at index 1 (or by name if preferred)
# Make a temporary copy to preserve original values for printing
X_train_original_for_example = X_train.copy()
original_area_value = X_train_original_for_example['area'].iloc[0]

# Learn from training data and scale it
X_train_scaled = scaler.fit_transform(X_train)
X_train = pd.DataFrame(X_train_scaled, columns=X_train.columns, index=X_train.index)

# Scale test data (just transform, don't use fit for output)
X_test_scaled = scaler.transform(X_test)
X_test = pd.DataFrame(X_test_scaled, columns=X_test.columns, index=X_test.index)

print("\nAfter scaling:")
print("All features now have similar range!")

# Example print using the 'area' column
if 'area' in X_train.columns:
    print(f"Example - First house 'area' before scaling: {original_area_value:.2f}")
    print(f"Example - First house 'area' after scaling: {X_train['area'].iloc[0]:.2f}")
else:
    print("Could not find 'area' column for example print.")
```

```python
#Train
from sklearn.linear_model import LogisticRegression

# Create model
model = LogisticRegression(max_iter=1000)

# Train the model
model.fit(X_train_scaled, y_train)
print("Model trained!")

# Check accuracy on training data
train_score = model.score(X_train_scaled, y_train)
print(f"Training accuracy: {train_score:.1%}")

# Check accuracy on test data
test_score = model.score(X_test_scaled, y_test)
print(f"Test accuracy: {test_score:.1%}")
```

```python
from sklearn.model_selection import cross_val_score

# Do 5-fold cross-validation
scores = cross_val_score(
    model,           # Our model
    X_train_scaled,  # Training data
    y_train,         # Training labels
    cv=5            # 5 mini-tests
)

print("5 mini-test scores:")
for i in range(5):
    print(f"  Test {i+1}: {scores[i]:.1%}")

print(f"\nAverage: {scores.mean():.1%}")
print(f"This means our model is {scores.mean():.1%} accurate!")
```
